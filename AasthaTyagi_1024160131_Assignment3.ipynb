## Q.6 Create a sample dataset (employees.csv)

Then determine the required statistics and transformations.


```python
employees = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
})

employees.to_csv('employees.csv', index=False)
employees
```

### a) Shape of the DataFrame


```python
print('Shape:', employees.shape)
print('Rows:', employees.shape[0], '| Columns:', employees.shape[1])
```

### b) Summary (dtypes and non-null counts)


```python
employees.info()
```

### c) Descriptive statistics


```python
employees.describe()
```

### d) First 5 rows and last 3 rows


```python
print('First 5 rows:')
display(employees.head())

print('Last 3 rows:')
display(employees.tail(3))
```

### e) Statistics
i. Average salary  
ii. Total bonus  
iii. Youngest employee's age  
iv. Highest performance rating


```python
print('i. Average salary:', employees['Salary'].mean())
print('ii. Total bonus:', employees['Bonus'].sum())
print('iii. Youngest employee age:', employees['Age'].min())
print('iv. Highest rating:', employees['Rating'].max())
```

### f) Sort by Salary (descending)


```python
employees.sort_values(by='Salary', ascending=False)
```

### g) Performance category column
- Excellent: rating >= 4.5  
- Good: 4.0 <= rating < 4.5  
- Average: rating < 4.0


```python
def performance_category(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

employees['Performance'] = employees['Rating'].apply(performance_category)
employees
```

### h) Identify missing values


```python
employees.isnull().sum()
```

### i) Rename Employee_ID to ID


```python
employees = employees.rename(columns={'Employee_ID': 'ID'})
employees
```

### j) Find employees who:
i. Have more than 5 years of experience  
ii. Belong to the IT department


```python
print('i. More than 5 years of experience:')
display(employees[employees['Years_of_Experience'] > 5])

print('\nii. IT department:')
display(employees[employees['Department'] == 'IT'])
```

### k) Add Tax column (10% of Salary)


```python
employees['Tax'] = employees['Salary'] * 0.10
employees
```

### l) Save modified DataFrame to a new CSV file


```python
employees.to_csv('employees_modified.csv', index=False)
print('Saved as employees_modified.csv')
pd.read_csv('employees_modified.csv')
```
