## Q.2 Questions on Basic NumPy Array


```python
# a) Reverse the array
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original:", arr)
print("Reversed:", arr[::-1])
```

```python
# b) Most frequent value and their indices

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def most_frequent(a):
    vals, counts = np.unique(a, return_counts=True)
    freq_val = vals[np.argmax(counts)]
    indices = np.where(a == freq_val)[0]
    return freq_val, indices

fx, ix = most_frequent(x)
print("x =", x)
print("Most frequent in x:", fx)
print("Indices:", ix)

print()
fy, iy = most_frequent(y)
print("y =", y)
print("Most frequent in y:", fy)
print("Indices:", iy)
```

## Q.4 linspace array named Aastha


```python
Aastha = np.linspace(10, 100, 25)
print("Array:", Aastha)
print("Dimensions:", Aastha.ndim)
print("Shape:", Aastha.shape)
print("Total elements:", Aastha.size)
print("Data type:", Aastha.dtype)
print("Total bytes:", Aastha.nbytes)

print("\nTranspose using reshape:")
print(Aastha.reshape(25, 1))

print("\nUsing .T attribute:")
print(Aastha.T)
print("For 1D array .T does not change it much, reshape works better for column vector.")
```

## Q.5 2-D Array ucs420_aastha


```python
ucs420_aastha = np.array([[10, 20, 30, 40],
                          [50, 60, 70, 80],
                          [90, 15, 20, 35]])
print("Original array:")
print(ucs420_aastha)

print("Mean:", np.mean(ucs420_aastha))
print("Median:", np.median(ucs420_aastha))
print("Max:", np.max(ucs420_aastha))
print("Min:", np.min(ucs420_aastha))
print("Unique elements:", np.unique(ucs420_aastha))

reshaped_ucs420_aastha = ucs420_aastha.reshape(4, 3)
print("\nReshaped (4x3):")
print(reshaped_ucs420_aastha)

resized_ucs420_aastha = np.resize(ucs420_aastha, (2, 3))
print("\nResized (2x3):")
print(resized_ucs420_aastha)
```
