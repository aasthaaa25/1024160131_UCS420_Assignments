## Q1. Sensor temperature readings


```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original:", temperature)

# a) add 2C using vectorization
corrected = temperature + 2
print("a) After adding 2:", corrected)

# b) C to F : F = (9/5)C + 32
fahrenheit = (9/5) * temperature + 32
print("b) Fahrenheit:", fahrenheit)

# c) readings > 32
above = temperature[temperature > 32]
print("c) Readings > 32:", above)

# d) count
print("d) Count > 32:", np.sum(temperature > 32))
```

**e)** Vectorization and boolean indexing are faster because NumPy works on the whole array at once in C, instead of looping one value at a time in Python.


## Q2. Daily steps of 4 users


```python
steps = np.array([[5000, 6200, 7100],
                  [8000, 7500, 9000],
                  [4500, 5100, 4800],
                  [9000, 8500, 9500]])
print(steps)

print("a) Total steps:", np.sum(steps))
print("b) Mean steps:", np.mean(steps))
print("c) Max:", np.max(steps), "Min:", np.min(steps))
print("d) Total per day (axis=0):", np.sum(steps, axis=0))
print("e) Total per user (axis=1):", np.sum(steps, axis=1))

idx = np.argmax(steps)
row = idx // steps.shape[1]
col = idx % steps.shape[1]
print("f) Max steps position -> user:", row, "day:", col, "value:", steps[row, col])
```

## Q3. Slicing, copy, reshape, flatten/ravel


```python
# a
original = np.array([1, 2, 3, 4, 5, 6])
print("original:", original)

# b
subset = original[1:5]
print("subset:", subset)

# c
subset[0] = 999
print("After changing subset[0] to 999:")
print("original:", original)
print("subset:", subset)
print("slice shares memory with original")
```

```python
# d - using copy
original = np.array([1, 2, 3, 4, 5, 6])
copied = original[1:5].copy()
copied[0] = 500
print("original:", original)
print("copied:", copied)
print("copy does not change original")
```

```python
# e
mat = np.arange(1, 13).reshape(3, 4)
print("3x4 matrix:")
print(mat)

# f
print("First row:", mat[0])
print("Last row:", mat[-1])
print("Second column:", mat[:, 1])
print("Rows 1-2, cols 2-3:")
print(mat[1:3, 2:4])
```

```python
# g, h, i
mat = np.arange(1, 13).reshape(3, 4)
print("Original mat:")
print(mat)

flat = mat.flatten()
rav = mat.ravel()
print("flatten:", flat)
print("ravel:", rav)

rav[0] = 100
print("\nAfter changing ravel[0]=100:")
print("mat:", mat)
print("ravel affects original")

mat = np.arange(1, 13).reshape(3, 4)
flat = mat.flatten()
flat[0] = 200
print("\nAfter changing flatten[0]=200:")
print("mat:", mat)
print("flatten does not affect original")

# j
print("\nshape:", mat.shape)
print("ndim:", mat.ndim)
print("size:", mat.size)
print("dtype:", mat.dtype)
```

## Q4. Assistance score prediction (OLS)


```python
X = np.array([[6, 70, 3],
              [5, 50, 6],
              [8, 80, 2],
              [4, 30, 8]])
y = np.array([40, 65, 30, 85])

print("a) Shape of X:", X.shape)
print("Dimensions:", X.ndim)

print("\nb) X.T:")
print(X.T)
print("Transpose flips rows and columns. Now each row is one feature.")

print("\nc) X.T @ X:")
print(X.T @ X)

XTX_inv = np.linalg.inv(X.T @ X)
print("\nd) Inverse of X.T @ X:")
print(XTX_inv)

# e) beta = (X^T X)^-1 X^T y
beta = XTX_inv @ X.T @ y
print("\ne) Coefficients beta:", beta)

print("\nf) beta[0] is for sleep hours, beta[1] for activity level, beta[2] for stress level.")
print("They show how much each feature affects the assistance score.")

# g
new_user = np.array([5, 40, 7])
pred = new_user @ beta
print("\ng) Predicted score for new_user:", pred)
```
