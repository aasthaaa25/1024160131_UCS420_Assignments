```python
import pandas as pd
import numpy as np
```

## 1. Load dataset (read_csv)


```python
url = "https://raw.githubusercontent.com/uiuc-cse/data-fa14/gh-pages/data/iris.csv"
df = pd.read_csv(url)
print("Shape:", df.shape)
df.head()
```

## 2. Explore data - info, describe, nulls, value_counts


```python
df.info()
print()
print("Missing values:")
print(df.isnull().sum())
print()
print("Species count:")
print(df["species"].value_counts())
df.describe()
```

## 3. Locate rows (loc / iloc)


```python
print("Rows 0, 4, 7, 8 using loc:")
display(df.loc[[0, 4, 7, 8]])
print()
print("Rows index 3 to 7:")
display(df.iloc[3:8])
print()
print("Rows 4 to 8, columns 1 to 3:")
display(df.iloc[4:9, 1:4])
```

## 4. Filtering conditions


```python
print("setosa only:")
display(df[df["species"] == "setosa"].head())
print()
print("petal_length greater than 1.5 and sepal_length greater than 5:")
display(df[(df["petal_length"] > 1.5) & (df["sepal_length"] > 5)].head(10))
```

## 5. NumPy array + basic ops


```python
num_cols = ["sepal_length", "sepal_width", "petal_length", "petal_width"]
arr = df[num_cols].to_numpy()
print("Array:")
print(arr[:5])
print("Shape:", arr.shape, "dtype:", arr.dtype)
print()
print("+2 on first 5 rows:")
print(arr[:5] + 2)
print()
print("*3:")
print(arr[:5] * 3)
print()
print("/2:")
print(arr[:5] / 2)
```

## 6. Reverse, unique, most frequent


```python
sample = arr[:10, 0]
print("Sample:", sample)
print("Reversed:", sample[::-1])
print("Unique values:", np.unique(sample))
species = df["species"].to_numpy()
vals, counts = np.unique(species, return_counts=True)
print()
print("Species:", vals)
print("Counts:", counts)
print("Most frequent species:", vals[np.argmax(counts)])
```

## 7. 2D access + reshape / resize


```python
print("Element row0 col1:", arr[0, 1])
print("Element row2 col0:", arr[2, 0])
vals12 = arr.ravel()[:12]
reshaped = vals12.reshape(4, 3)
print()
print("Reshape 4x3:")
print(reshaped)
resized = np.resize(vals12, (2, 3))
print()
print("Resize 2x3:")
print(resized)
```

## 8. Mean median max min std + axis


```python
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Max:", np.max(arr), "Min:", np.min(arr))
print("Std:", np.std(arr))
print()
print("Column-wise mean axis=0:", np.mean(arr, axis=0))
print("Row-wise mean first 5 axis=1:", np.mean(arr, axis=1)[:5])
print("Column-wise sum:", np.sum(arr, axis=0))
```

## 9. Vectorization + boolean indexing


```python
petal = df["petal_length"].to_numpy()
scaled = petal * 1.1
print("Original first 5:", petal[:5])
print("Scaled *1.1:", scaled[:5])
mask = petal > 4
print()
print("petal_length > 4 (first 10):", petal[mask][:10])
print("Count > 4:", np.sum(mask))
print("Mean of those:", np.mean(petal[mask]))
```

## 10. Slice vs copy, flatten vs ravel


```python
original = arr[0].copy()
print("original row0:", original)
subset = original[1:4]
subset[0] = 999
print("after slice change, original:", original)
original = arr[0].copy()
copied = original[1:4].copy()
copied[0] = 500
print("after copy change, original:", original)
mat = arr.ravel()[:12].reshape(3, 4).copy()
print()
print("3x4 matrix:")
print(mat)
flat = mat.flatten()
rav = mat.ravel()
print("flatten:", flat)
print("ravel:", rav)
rav[0] = -1
print("after ravel change, mat:")
print(mat)
print("(ravel shares memory)")
mat2 = arr.ravel()[:12].reshape(3, 4).copy()
flat2 = mat2.flatten()
flat2[0] = -2
print("after flatten change, mat2:")
print(mat2)
print("(flatten is copy)")
```

## 11. Add columns, np.where, rename, sort


```python
df["petal_area"] = df["petal_length"] * df["petal_width"]
df["sepal_area"] = df["sepal_length"] * df["sepal_width"]
df["petal_size"] = np.where(df["petal_area"] > df["petal_area"].mean(), "Large", "Small")
df = df.rename(columns={"species": "Species"})
print(df.head())
print()
print("Sorted by petal_area:")
display(df.sort_values(by="petal_area", ascending=False).head())
```

## 12. groupby


```python
print(df.groupby("Species")[num_cols].mean())
print()
print(df.groupby("Species")["petal_area"].agg(["mean", "max", "min", "count"]))
print()
print("petal_size counts:")
print(df["petal_size"].value_counts())
```

## 13. drop row/column + save csv


```python
temp = df.copy()
temp = temp.drop(index=4)
temp = temp.drop(columns=["sepal_area"])
print("After delete row4 and sepal_area:", temp.shape)
display(temp.head())
temp.to_csv("1024160131_iris_practice.csv", index=False)
print("Saved: 1024160131_iris_practice.csv")
```

## 14. argmax + correlation


```python
idx = np.argmax(df["petal_area"].to_numpy())
print("Max petal_area index:", idx)
print(df.loc[idx])
print()
print("Correlation:")
print(df[num_cols + ["petal_area"]].corr())
```
